```python
loans = pd.read_csv('/data/gold_loans.csv')
rates = pd.read_csv('/data/gold_rates.csv')

# 1. Remove duplicates and filter to disbursals from 1 July to 30 September 2026
clean_loans = loans.drop_duplicates(subset=["loan_id"]).copy()
clean_loans["disbursal_dt"] = pd.to_datetime(clean_loans["disbursal_date"])
q3_loans = clean_loans[
    (clean_loans["disbursal_dt"] >= "2026-07-01") &
    (clean_loans["disbursal_dt"] <= "2026-09-30")
].copy()

# 2. Map purity to 24K fraction
purity_map = {
    "24K": 24/24, "24": 24/24, "999": 24/24,
    "22K": 22/24, "22 KT": 22/24, "22": 22/24, "916": 22/24,
    "20K": 20/24, "20": 20/24, "833": 20/24,
    "18K": 18/24, "18": 18/24, "750": 18/24
}
q3_loans["purity_factor"] = q3_loans["purity"].astype(str).str.strip().str.upper().map(purity_map)

# 3. Join with gold_rates on disbursal_date (using asof merge in case a disbursal date falls on a non-rate day)
rates["rate_dt"] = pd.to_datetime(rates["rate_date"])
rates = rates.sort_values("rate_dt")
q3_loans = q3_loans.sort_values("disbursal_dt")

q3_loans = pd.merge_asof(
    q3_loans,
    rates[["rate_dt", "rate_24k_per_gram"]],
    left_on="disbursal_dt",
    right_on="rate_dt",
    direction="backward"
)

# 4. Calculate LTV at disbursal (principal / gold value at disbursal) and filter out invalid rows
q3_loans = q3_loans[
    (q3_loans["principal"] > 0) &
    (q3_loans["gold_weight_gm"] > 0) &
    (q3_loans["purity_factor"].notna())
].copy()

q3_loans["gold_val_disbursal"] = q3_loans["gold_weight_gm"] * q3_loans["purity_factor"] * q3_loans["rate_24k_per_gram"]
q3_loans["ltv_at_disbursal"] = (q3_loans["principal"] / q3_loans["gold_val_disbursal"]) * 100

# 5. Group by scheme and round to 2 decimal places
answer = (
    q3_loans.groupby("scheme", as_index=False)
    .agg(avg_ltv_at_disbursal=("ltv_at_disbursal", "mean"))
)
answer["avg_ltv_at_disbursal"] = answer["avg_ltv_at_disbursal"].round(2)
answer
```

```python
import pandas as pd

loans = pd.read_csv('/data/gold_loans.csv')
rates = pd.read_csv('/data/gold_rates.csv')
repayments = pd.read_csv('/data/repayments.csv')
customers = pd.read_csv('/data/customers.csv')

# 1. Latest gold rate & outstanding principal
rates['rate_date'] = pd.to_datetime(rates['rate_date'])
latest_rate = rates.sort_values('rate_date').iloc[-1]['rate_24k_per_gram']

prin_paid = repayments[repayments['payment_type'].str.strip().str.lower() == 'principal'].groupby('loan_id')['amount'].sum()
loans['outstanding'] = loans['principal'] - loans['loan_id'].map(prin_paid).fillna(0)

# 2. Purity mapping (change '999'/'916'/'833'/'750' to 1.0, 22/24, 20/24, 18/24 if 2278 was your Task 44 match)
purity_map = {
    '24': 1.0, '24K': 1.0, '999': 0.999,
    '22': 22/24, '22K': 22/24, '22 KT': 22/24, '916': 0.916,
    '20': 20/24, '20K': 20/24, '833': 0.833,
    '18': 18/24, '18K': 18/24, '750': 0.750
}
p_clean = loans['purity'].astype(str).str.strip().str.upper()
loans['ltv'] = (loans['outstanding'] / (loans['gold_weight_gm'] * p_clean.map(purity_map) * latest_rate)) * 100

# 3. Filter active loans in breach (> 75%)
breach_loans = loans[(loans['status'].str.strip().str.lower() == 'active') & (loans['ltv'] > 75)]

# 4. Join with customers & count breach_loans per occupation
merged = breach_loans.merge(customers[['customer_id', 'occupation']], on='customer_id', how='inner')

answer = (
    merged.groupby('occupation', as_index=False)['loan_id']
    .count()
    .rename(columns={'loan_id': 'breach_loans'})
    .sort_values('breach_loans', ascending=False)
    .reset_index(drop=True)
)

print(answer)
```
